```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from scipy.special import expit
```

```python
data=load_breast_cancer()
```

```python
x=data.data
y=data.target
```

```python
y[y == 0] = -1
```

```python
df=pd.DataFrame(x, columns=data.feature_names)
df
```

```python
x_train, x_test, y_train, y_test=train_test_split(x,y, test_size=0.2, random_state=123)
```

```python
medias = x_train.mean(axis=0)
desv = x_train.std(axis=0)

x_train = (x_train - medias) / desv
x_test = (x_test - medias) / desv
```

```python
w=np.zeros(x_train.shape[1])
b=0
ciclos=500
rate=0.2
historial=[]
costos=[]
for ciclo in range(ciclos):
  costo=0
  errores=0
  indices=np.random.permutation(len(x_train))
  for i in indices:
    xs=x_train[i]
    t=y_train[i]
    a=np.dot(w,xs)+b
    if a*t<=0:
      w=w+rate*t*xs
      b=b+rate*t
      errores+=1
      costo-=a*t
    else:
      pass
  historial.append(errores)

  costo=0
  for xs, t in zip(x_train, y_train):

        a = np.dot(w, xs) + b

        if a*t <= 0:

            costo += -a*t

  costos.append(costo)

```

```python
#PREDICCIONES
y_pred=np.dot(w,x_test.T)+b
accuracy_score(y_test,np.sign(y_pred))
```

```python
#REGRESSION LOGISTICA MEDIANTE GRADIENTE ESTOCASTICO DESCENDIENTE
y[y==-1]=0
y_train[y_train==-1]=0
y_test[y_test==-1]=0
```

```python
epoch=500
w=np.zeros(x_train.shape[1])
b=0
rate=0.1

maxllh=[]
for ciclos in range(epoch):
  llh=0
  indice=np.random.permutation(len(x_train))
  for i in indice:
    xs=x_train[i]
    t=y_train[i]
    y=1/(1+np.exp(-(np.dot(w,xs)+b)))
    y = np.clip(y, 1e-15, 1-1e-15)
    grad=(y-t)*xs
    w-=rate*grad
    b-=rate*(y-t)

  for xs, t in zip(x_train,y_train):
    z=np.dot(w,xs)+b
    s=1/(1+np.exp(-z))
    s=np.clip(s, 1e-15, 1-1e-15)
    llh+=t*np.log(s)+(1-t)*np.log(1-s)
  maxllh.append(llh)



```

```python
class BayesLogistica:
  def __init__(self, alpha=1):
    #prior
    self.alpha=alpha
  def fit(self, Phi, t, max_iter=100, tol=1e-3):
    N,M = Phi.shape
    w=np.zeros(M)
    I=np.eye(M)
    for i in range(max_iter):
      y=expit(Phi@w)
      g=Phi.T@(y-t) + self.alpha*w
      R=np.diag(y*(1-y))
      H=Phi.T@R@Phi + self.alpha*I
      step=np.linalg.solve(H,g)
      wnew=w-step
      if np.linalg.norm(w-wnew)<tol:
        w=wnew
        break
      w=wnew
    self.w_map=w
    self.S=np.linalg.inv(H)
    return self

  def latent_distribution(self,phi):
    mu=phi@self.w_map
    sigma=phi@self.S@phi
    return mu, sigma

  def predict_proba(self,Phi):
    prob=[]
    for phi in Phi:
      mu, sigma=self.latent_distribution(phi)
      k=(1+np.pi*sigma/8)**(-1/2)
      proba=expit(mu*k)
      prob.append(proba)
    return np.array(prob)






```

```python
modelo=BayesLogistica()
modelo.fit(x_train, y_train)
```
